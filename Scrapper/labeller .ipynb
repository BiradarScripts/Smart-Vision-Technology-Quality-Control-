```python
import os
import json
import requests
import base64
from PIL import Image
import cv2

# Correct Groq API URL and API Key
GROQ_API_URL = "https://api.groq.com/v1/tasks/brand-detection"  # Replace with the correct endpoint
GROQ_API_KEY = "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"  # Replace with your valid API key

def detect_brand_with_groq(image_path):
    """
    Detects brand name using Groq API and provides bounding boxes.
    Args:
        image_path (str): Path to the input image
    Returns:
        List[dict]: List of detected brand names with bounding boxes
    """
    headers = {
        "Authorization": f"Bearer {GROQ_API_KEY}",
        "Content-Type": "application/json"
    }

    # Prepare image data
    with open(image_path, "rb") as f:
        image_base64 = base64.b64encode(f.read()).decode("utf-8")
    
    # Groq API payload
    payload = {
        "task": "detect_brand",
        "image": image_base64,  # Use base64-encoded image
        "parameters": {
            "text_prompt": "Detect and provide bounding boxes for brand names in this image."
        }
    }

    response = requests.post(GROQ_API_URL, headers=headers, json=payload)
    if response.status_code == 200:
        return response.json()  # Expected to return brand data as JSON
    else:
        print(f"Error: {response.status_code} - {response.text}")
        return []

def draw_bounding_boxes(image_path, brand_data, output_folder):
    """
    Draw bounding boxes on the image for detected brand names.
    Args:
        image_path (str): Path to the input image
        brand_data (list): Detected brand bounding boxes and text
        output_folder (str): Path to save the output images
    """
    image = cv2.imread(image_path)
    for item in brand_data:
        x, y, w, h = item["bounding_box"]  # Coordinates for bounding box
        brand_name = item["brand"]
        cv2.rectangle(image, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv2.putText(image, brand_name, (x, y-10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
    
    # Save output image
    image_name = os.path.basename(image_path)
    output_path = os.path.join(output_folder, f"output_{image_name}")
    cv2.imwrite(output_path, image)
    print(f"Output saved at {output_path}")

def process_images_in_folder(input_folder, output_folder):
    """
    Process all images in the input folder and save the outputs with bounding boxes.
    Args:
        input_folder (str): Folder containing input images
        output_folder (str): Folder to save the output images
    """
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # Iterate through all files in the input folder
    for filename in os.listdir(input_folder):
        if filename.lower().endswith((".jpg", ".jpeg", ".png")):  # Filter image files
            image_path = os.path.join(input_folder, filename)
            print(f"Processing {image_path}...")
            try:
                brand_data = detect_brand_with_groq(image_path)
                if brand_data:
                    draw_bounding_boxes(image_path, brand_data, output_folder)
                else:
                    print(f"No brand data found for {image_path}.")
            except Exception as e:
                print(f"Error processing {image_path}: {e}")

# Example usage
if __name__ == "__main__":
    input_folder = "./Deo"  # Replace with your input folder
    output_folder = "./labelout"  # Replace with your output folder
    process_images_in_folder(input_folder, output_folder)
```
